### Creating tensors from arrays

```python
# creates a copy of the numpy array!
t_b = torch.tensor(b, dtype=torch.float32)
print(t_b)
```

```python
# creates a copy of the numpy array!
t_b = torch.FloatTensor(b)
print(t_b)
```

```python
torch.from_numpy?
```

```python
# tensor inherits numpy array dtype!
t_b = torch.from_numpy(b)
print(t_b)
```

```python
# no copy of the array is made, so changes to array also change tensor!
b[3] = -99.0
print(b)
print(t_b)
```

## Applying mathematical operators to tensors

```python
# another way to create a random generator
prng = torch.manual_seed(1234)
```

```python
# itemwise binary operations like +-*/ work as expected
t7 = 2 * torch.rand(5, 2, generator=prng) - 1

print(t7)
```

```python
# other distributions are available
t8 = torch.normal(mean=0, std=1, size=(5,2), generator=prng)

print(t8)
```

```python
# itemwise multiplication of tensors
torch.multiply(t7, t8)
```

```python
t7.multiply(t8)
```

```python
t7 * t8
```

### Exercise

Create two random normal tensors `y_true` and the other `y_pred` with shape 60,000x1. Apply the exponential function to both tensors. Compute the the root mean squared log error between the two tensors. What is the shape of the resulting tensor?

#### Solution

```python
y_true = torch.exp(torch.normal(mean=0, std=1, size=(60_000, 1)))
y_pred = torch.exp(torch.normal(mean=0, std=1, size=(60_000, 1)))

rmsle = torch.mean((torch.log(y_true) - torch.log(y_pred))**2, dim=0)**0.5

print(rmsle)
print(rmsle.shape)
```

## Chunk, split, stack, and concatenate tensors

```python
prng = torch.manual_seed(42)
```

```python
dataset = torch.rand(10, 3, generator=prng)
print(dataset)
```

```python
torch.chunk?
```

```python
batches = torch.chunk(dataset, 3, dim=0)
```

```python
for i, batch in enumerate(batches):
    print(batch)
```

```python
features = torch.chunk(dataset, 3, dim=1)

for j, feature in enumerate(features):
    print(feature)
```

```python
splits = torch.split(dataset, split_size_or_sections=[4, 4, 2], dim=0)
```

```python
for i, split in enumerate(splits):
    print(split)
```

```python
torch.stack?
```

```python
a = torch.zeros(5)
b = torch.ones(5)

print(a.shape)
print(b.shape)
```

```python
c = torch.stack([a, b], dim=0)

print(c)
print(c.shape)
```

```python
c = torch.stack([a, b], dim=1)

print(c)
print(c.shape)
```

```python
torch.cat?
```

```python
c = torch.cat([a, b], dim=0)

print(c)
print(c.shape)
```

```python
torch.cat([a, b], dim=1)
```
